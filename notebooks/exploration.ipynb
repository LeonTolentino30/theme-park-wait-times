```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("../data/raw/wait_times.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
sns.barplot(data=df, x="ride", y="wait_time")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.lineplot(data=df, x="hour", y="wait_time")
plt.show()
```

```python
sns.barplot(data=df, x="day", y="wait_time")
plt.show()
```

```python
df.isna().sum()
```

```python
df['wait_time'] = df['wait_time'].fillna(df['wait_time'].median())
```

```python
df['is_peak'] = df['hour'].between(11, 17)
```

```python
df['is_weekend'] = df['day'].isin(['Saturday', 'Sunday'])
```
